# Problem 1

```python
L = [1,2,3,4,5]

sum=0

for i in L:
    sum = sum +i
print(sum)

product =1
for i in L:
    product = product*i
print(product)
```

- O(n) + O(n) = O(2n) = O(n)
- Order of growth is linear

# Problem 4

```python
def intToStr(i):
    digits = '0123456789'
    if i==0:
        return '0'
    result = ''
    while i>0:
        result = digits[i%10] + result
        print(result) #add print to see what's going on under the hood
        i = i//10

    return result

print(intToStr(123))
```
