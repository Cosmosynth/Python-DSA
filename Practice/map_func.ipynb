```python
def sqr(x):
    return x*x

sqr(2)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

list(map(sqr, numbers))
```

```python
## Lambda func with map

list(map(lambda x:x*x, numbers))
```

```python
# Map multiple iterables

number1 = [1, 3, 4]
number2 = [1, 2 ,5]

added_num = list(map(lambda x,y: x + y, number1 , number2))
```

```python
str_num = ["1", "2", "3", "4"]
int_num = list(map(int, str_num))
print(int_num)
```

```python
words = ["apple", "banana", "mango"]
uppercase = list(map(str.upper, words))
print(uppercase)
```

```python
def get_name(person):
    return person["name"]

people = [
    {"name":"Hasya", "age":"17"},
    {"name":"Shr", "age": "17"}
]


list(map(get_name, people))
```
